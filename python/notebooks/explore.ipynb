```python
from anthropic import Anthropic

from askkit.ask import answer_text, ask
from askkit.config import load_api_key
from askkit.prompts import ASK_SYSTEM

client = Anthropic(api_key=load_api_key())
```

```python
response = ask(
    client,
    "How does a context window work?",
    "claude-haiku-5-5",
    1000,
    ASK_SYSTEM,
)
```

```python
print(answer_text(response))
```
